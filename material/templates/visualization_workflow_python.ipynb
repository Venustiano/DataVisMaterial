## 4. Prepare the data

### Ask AI 


### Copy the code returned by AI in the next cell and run it

````python
import pandas as pd

# 1. Working copy: the original df stays untouched
data = df.copy()

# 2. Checks (nothing is changed here)
print("Missing values per column:")
print(data.isna().sum(), "\n")
print("Fully duplicated rows:", data.duplicated().sum())
print("Duplicated year-month pairs:", data.duplicated(subset=["year", "month"]).sum())
print("\nNumber of distinct months per year (expected 12):")
print(data.groupby("year")["month"].nunique(), "\n")

# 3. Check the month labels before converting them
month_order = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
               "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]
unexpected = set(data["month"].unique()) - set(month_order)
print("Unexpected month labels:", unexpected if unexpected else "none")

# 4. Make month an ordered category (values stay the same, only the order is defined)
data["month"] = pd.Categorical(data["month"], categories=month_order, ordered=True)
print("Missing months after conversion:", data["month"].isna().sum(), "(should be 0)")

# 5. Sort chronologically
data = data.sort_values(["year", "month"]).reset_index(drop=True)

display(data.head())
print("Working dataset dimensions:", data.shape)   # should still be (24, 3)
````

## 6. Create the first visualization

Ask AI to generate a small, understandable code cell using the actual column names in your dataset. Ask it to explain each important line and state any assumptions.

### Ask AI
Prompt idea: *“Write Python code for the chart we selected. Use my existing dataframe named `data` and these exact columns: [insert names]. Use matplotlib. Include a descriptive title, axis labels with units, and an appropriate legend if needed. Explain the code and flag assumptions.”*

````python
import matplotlib.pyplot as plt
import pandas as pd

# Define the correct chronological order for month abbreviations
month_order = [
    'Jan',
    'Feb',
    'Mar',
    'Apr',
    'May',
    'Jun',
    'Jul',
    'Aug',
    'Sep',
    'Oct',
    'Nov',
    'Dec',
]

# Ensure 'month' is treated as an ordered categorical variable
# so matplotlib displays them chronologically rather than alphabetically
data['month'] = pd.Categorical(
    data['month'], categories=month_order, ordered=True
)

# Sort the dataset by year and the ordered month
df_sorted = data.sort_values(['year', 'month'])

# Initialize the figure canvas
plt.figure(figsize=(10, 6))

# Loop through each unique year and plot its respective line
for year, group in df_sorted.groupby('year'):
  plt.plot(
      group['month'],
      group['mean_temp_c'],
      marker='o',  # Adds clear markers for each monthly data point
      linewidth=2,
      label=str(year),
  )

# Add descriptive title and axis labels with units
plt.title(
    'Monthly Mean Temperature Variation and Comparison Across Years',
    fontsize=14,
    pad=15,
)
plt.xlabel('Month', fontsize=12)
plt.ylabel('Mean Temperature (°C)', fontsize=12)

# Customize legend, grid, and layout
plt.legend(title='Year', loc='upper right', frameon=True)
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()

# edit the filename and run after creating the final chart
# Option 
plt.savefig("final_visualization.png", dpi=300, bbox_inches="tight")

# Display the plot
plt.show()
````
