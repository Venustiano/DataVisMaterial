## 1. Define the research question

**Research question:** How does the monthly mean temperature vary throughout the year, and how do the patterns differ between the example years?

**Visualization objective:** Compare monthly temperature patterns across years.

**Audience:** Students and researchers who need a quick comparison of seasonal patterns.


## 2. Load the dataset

The CSV file `example_temperature.csv` should be in the same folder as this notebook. It contains one row per month and year.

- `year`: example year
- `month`: calendar month
- `mean_temp_c`: illustrative monthly mean temperature in degrees Celsius


````python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("example_temperature.csv")
display(df.head())
````

## 3. Explore the dataset

Check its size, columns, data types, and summary statistics before plotting.


````python
print("Rows and columns:", df.shape)
print("Columns:", df.columns.tolist())
df.info()
display(df.describe())
display(df.isna().sum().to_frame("missing_values"))
````

**Exploration notes:** Each row represents one month in one year. The temperature variable is numeric and measured in degrees Celsius. This small teaching dataset has no missing values. The values are illustrative, not empirical.


## 4. Prepare the data

For this simple example, no filtering or transformation is necessary. We make a working copy and ensure the month order is chronological.


````python
data = df.copy()

month_order = [
    "Jan", "Feb", "Mar", "Apr", "May", "Jun",
    "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"
]
data["month"] = pd.Categorical(
    data["month"], categories=month_order, ordered=True
)
data = data.sort_values(["year", "month"])

display(data.head())
````

## 5. Choose a visualization

A line chart is appropriate because months have a natural order and the objective is to compare the annual pattern over time. Each line represents one year.

A line chart suggests continuity between monthly values, so readers should understand that these are monthly summaries rather than daily measurements.


## 6. Create the visualization

The code below reshapes the data so each year becomes a separate line, then plots monthly mean temperature.


````python
wide = data.pivot(index="month", columns="year", values="mean_temp_c")

ax = wide.plot(marker="o", figsize=(10, 5))
ax.set_title("Monthly mean temperature by year (illustrative data)")
ax.set_xlabel("Month")
ax.set_ylabel("Mean temperature (°C)")
ax.legend(title="Year")
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
````

## 7. Evaluate and refine the visualization

Consider:
- Are the months in the correct order?
- Are the units explicit?
- Can the reader distinguish the years?
- Is the title clear that the data are illustrative?
- Would direct labels or a different layout improve accessibility?

For this example, the month order, units, and legend are explicit. In a real research setting, also consider uncertainty, measurement methods, missing months, and whether the years are comparable.


## 8. Interpret the results

**Example interpretation:** The illustrative data show a seasonal cycle, with lower mean temperatures near the beginning and end of the year and higher values around mid-year. The two example years follow broadly similar patterns, but their monthly values differ.

**Limitations:** These are fictional values for demonstrating a workflow. They cannot support conclusions about actual climate, climate change, or any real location. Two years would also be insufficient to establish a long-term climate trend.
