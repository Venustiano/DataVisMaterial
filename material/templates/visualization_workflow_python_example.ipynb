# 4. Prepare the data

🟦 **COPY → AI** Copy the prompt in the raw cell below, replace the placeholder between the triple quotes with the **output of the exploration cell in section 3**, and send it to your AI chat.

🟩 Then paste the code returned by the AI in the marked code cell, read it, and run it.

# 🟩 **PASTE ← AI** Copy the code returned by AI in the next cell and run it

````python
import pandas as pd

# 1. Working copy: the original df stays untouched
data = df.copy()

# 2. Checks (nothing is changed here)
print("Missing values per column:")
print(data.isna().sum(), "\n")
print("Fully duplicated rows:", data.duplicated().sum())
print("Duplicated year-month pairs:", data.duplicated(subset=["year", "month"]).sum())

# 3. Make month an ordered category so plots are chronological
month_order = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
               "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]
data["month"] = pd.Categorical(data["month"], categories=month_order, ordered=True)

# 4. Sort chronologically
data = data.sort_values(["year", "month"]).reset_index(drop=True)

display(data.head())
print("Working dataset dimensions:", data.shape)
````

# 5. Choose a visualization

🟨 **STUDENT INPUT** — decide what relationship or comparison you need to show before choosing a chart.

- **What are the variables?** `month` (time), `year` (group), `mean_temp_c` (quantity, °C).
- **What role does each variable play?** (time, category, quantity, group, location, etc.): month: ordered time; year: grouping factor; mean_temp_c: numeric value.
- **What comparison, pattern, or relationship do you want readers to see?** The seasonal cycle per year, and how the two years differ month by month.
- **What chart type do you propose, and why?** Overlaid line chart: months have a natural order and the goal is to compare annual patterns; each year is a line.

# 6. Create the first visualization

🟦 **COPY → AI** Edit the prompt in the raw cell below: fill in the chart type you chose in section 5 and your exact column names, then send it to your AI chat.

🟩 **PASTE ← AI** Paste the code returned by the AI in the marked code cell below, read it, and run it. If it fails, copy the error message and the code into your AI chat and ask for a fix — then check the fix before running it.

````python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

for year, group in data.sort_values(["year", "month"]).groupby("year"):
    plt.plot(group["month"], group["mean_temp_c"],
             marker="o", linewidth=2, label=str(year))

plt.title("Monthly mean temperature by year (illustrative data)", fontsize=14, pad=15)
plt.xlabel("Month", fontsize=12)
plt.ylabel("Mean temperature (°C)", fontsize=12)
plt.legend(title="Year", loc="upper right")
plt.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
````
