## 1. Define the research question

**Research question:** How does monthly mean temperature vary throughout the year, and how do patterns differ between the example years?

**Objective:** Compare monthly temperature patterns across years.

**Audience:** Students and researchers interested in seasonal patterns.


## 2. Load the dataset

Place `example_temperature.csv` in the same folder as this notebook.

- `year`: example year
- `month`: calendar month
- `mean_temp_c`: illustrative monthly mean temperature in degrees Celsius


````r
library(readr)
library(ggplot2)

data <- read_csv("example_temperature.csv")
head(data)
````

## 3. Explore the dataset

Check dimensions, columns, types, summary statistics, and missing values.


````r
dim(data)
names(data)
str(data)
summary(data)
sapply(data, function(x) sum(is.na(x)))
````

**Exploration notes:** Each row represents one month in one year. Temperature is numeric and measured in degrees Celsius. The teaching dataset has no missing values. Values are illustrative, not empirical.


## 4. Prepare the data

No filtering or transformation is necessary for this example. Set the month factor levels so that months appear in calendar order.


````r
month_order <- c(
  "Jan", "Feb", "Mar", "Apr", "May", "Jun",
  "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"
)

analysis_data <- data
analysis_data$month <- factor(
  analysis_data$month,
  levels = month_order,
  ordered = TRUE
)
````

## 5. Choose a visualization

A line chart is suitable because months have a natural order and the objective is to compare annual patterns. Each line represents one year. The lines connect monthly summaries; they are not daily measurements.


## 6. Create the visualization

Use `geom_line()` to connect months and `geom_point()` to show the monthly observations.


````r
final_plot <- ggplot(
  analysis_data,
  aes(x = month, y = mean_temp_c, color = factor(year), group = year)
) +
  geom_line(linewidth = 1) +
  geom_point(size = 2) +
  labs(
    title = "Monthly mean temperature by year (illustrative data)",
    x = "Month",
    y = "Mean temperature (°C)",
    color = "Year"
  ) +
  theme_minimal()

print(final_plot)
````

## 7. Evaluate and refine

Check that:
- Months appear in calendar order.
- Temperature units are explicit.
- Years can be distinguished.
- The title clearly identifies the data as illustrative.

For real research, consider measurement methods, uncertainty, missing months, and comparability between years.


## 8. Interpret the results

**Example interpretation:** The illustrative data show a seasonal cycle, with lower mean temperatures near the beginning and end of the year and higher values around mid-year. The two example years follow broadly similar patterns, but their monthly values differ.

**Limitations:** These are fictional values. They cannot support conclusions about actual climate, climate change, or a real location. Two years are also insufficient to establish a long-term climate trend.
